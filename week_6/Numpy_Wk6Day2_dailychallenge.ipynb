```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import time
```

# Step 1: generate the data

```python
# === LIVE CODING: Generate synthetic temperature data ===
# np.random.seed(42)
# city_temps = np.random.uniform(-5, 35, (10, 12))
# city_names = [...]
# month_names = [...]

np.random.seed(42)
city_temps = np.random.uniform(-5, 35, (10, 12)).round(1)
city_names = ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix", "Philadelphia", "San Antonio", "San Diego", "Dallas", "San Jose"]
month_names = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]

print (city_temps)
print (city_temps.shape)
```

# Step 2: Analyze with NumPy

```python
# === LIVE CODING: Analyze ===
# Annual average per city: .mean(axis=1)
# Hottest/coldest city: .argmax(), .argmin()


avg_city = city_temps.mean(axis=1)
print(avg_city)
hottest_city = city_names[avg_city.argmax()]
print(f'Hottest city: {hottest_city}, Average: {avg_city[avg_city.argmax()].round(1)}')
coldest_city = city_names[avg_city.argmin()]
print(f'Coldest city: {coldest_city}, Average: {avg_city[avg_city.argmin()].round(1)}')

# Overall hottest/coldest month: .max(), .min()
```

## Step 3: Convert to DataFrame

```python
weather_data = pd.DataFrame(city_temps, index=city_names, columns=month_names)
weather_data
```

## Step 4: Visualize

### Annual Average Temperatures per City

```python
plt.figure(figsize=(12, 6))
sns.barplot(x=weather_data.index, y=weather_data.mean(axis=1), palette='viridis')
plt.title('Annual Average Temperature per City', fontsize=25)
plt.xlabel('City', fontsize=15)
plt.ylabel('Average Temperature (°C)', fontsize=15)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Monthly Temperature Trends for All Cities

```python
plt.figure(figsize=(14, 8))
for city in city_names:
    sns.lineplot(x=month_names, y=weather_data.loc[city], label=city)
plt.title('Monthly Temperature Trends for All Cities', fontsize = 25)
plt.xlabel('Month', fontsize=15)
plt.ylabel('Temperature (°C)', fontsize=15)
plt.xticks(rotation=45, ha='right')
plt.grid(linestyle='--', alpha=0.7)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
