### Problem Statement and Objective
The primary objective of this project is to develop a binary classification model that predicts the target variable `loan_default` (where 1 indicates a borrower will fail to meet contractual repayment obligations within a specified 12-month window, and 0 indicates full repayment) to support automated and risk-tiered credit underwriting decisions.

### Required Data Types
* **Credit History and Scoring Data:** Historical FICO or credit scores, previous delinquency records, and derogatory marks are required because an applicant's past repayment behavior serves as the strongest statistical indicator of future default probability.
* **Applicant Financial and Income Profiles:** Documented gross annual income, employment status, and employment tenure are necessary because steady cash flow demonstrates the borrower's capacity to service regular debt obligations.
* **Loan Characteristics:** Key parameters including requested principal loan amount, loan term length, interest rate, and borrowing purpose must be collected because they dictate the borrower's monthly debt burden relative to their overall financial capacity.
* **Existing Liabilities and Leverage:** Total outstanding balances, credit utilization rates, and the debt-to-income (DTI) ratio are critical because excessive leverage significantly magnifies the risk of insolvency under financial stress.
* **Internal Banking and Transaction Records:** Average account balances, savings deposits, and historical overdraft frequencies are valuable because granular transaction history provides direct insight into day-to-day liquidity management and baseline fiscal reliability.

### Realistic Data Sources and Integration
* **National Credit Bureaus (e.g., Experian, Equifax, TransUnion):** Integrate secure RESTful APIs to ingest standardized credit reports, credit inquiries, and public financial records in real time during the initial application flow.
* **Core Banking and Account Servicing Databases:** Extract existing customer deposit histories, active accounts, and past loan repayment records via direct internal SQL database queries and scheduled ETL pipelines.
* **Customer Application Portals:** Ingest self-reported applicant details, stated loan purposes, and uploaded identity documents through front-end digital origination web forms and secure file upload endpoints.
* **Third-Party Payroll and Asset Verification Services (e.g., Plaid, Argyle):** Implement third-party API connectors to securely authenticate bank accounts and automatically verify declared income and employment directly from payroll providers.

### Risks, Constraints, and Governance
Deploying a predictive credit model involves substantial regulatory, ethical, and operational constraints. Under lending frameworks like the Equal Credit Opportunity Act (ECOA) and Fair Credit Reporting Act (FCRA), models must exclude legally protected demographic characteristics (such as race, gender, and marital status) and provide clear, compliant adverse action notices explaining the reasons behind credit denials. Stringent privacy regulations, including GDPR and CCPA, mandate rigorous data anonymization, robust encryption at rest and in transit, and role-based data access controls. Furthermore, historical credit data is inherently vulnerable to sampling bias, as records only reflect applicants previously approved by earlier underwriting policies; training exclusively on this subset can introduce selection bias and perpetuate past exclusionary practices. To mitigate these risks, ongoing model governance requires continuous validation to prevent disparate impact, proactive monitoring for concept drift against changing macroeconomic conditions, and strict data quality audits to detect anomalies or pipeline failures before predictions are served.

### Your answer

**Predicting Stock Prices:** This scenario is most suitable for **Supervised Learning**, specifically a **Regression** task. The objective is to predict a continuous numerical value (future stock price).

*   **Input Data:** The input data would consist of historical stock prices (e.g., open, close, high, low), trading volume, company financial reports, news sentiment, and relevant macroeconomic indicators.
*   **Output:** The model would output a numerical value representing the predicted stock price at a future point in time.
*   **Learning Signal/Objective:** The learning signal is the actual future stock price, and the objective is to minimize the difference (error) between the predicted and actual stock prices, typically measured by metrics like Mean Squared Error (MSE) or Root Mean Squared Error (RMSE).

**Organizing a Library of Books:** This scenario is best addressed by **Unsupervised Learning**, specifically **Clustering**. The goal is to discover inherent groupings within the data without prior knowledge of labels.

*   **Input Data:** The input data would include textual information such as book titles, author names, book descriptions, genre tags, and potentially publication dates or reader ratings.
*   **Output:** The output would be clusters of books, where each cluster represents a distinct genre or category, and books within a cluster share similar characteristics.
*   **Learning Signal/Objective:** There is no explicit learning signal; instead, the objective is to maximize the similarity of books within the same cluster while minimizing the similarity between books in different clusters, often using distance metrics.

**Program a robot to navigate and find the shortest path in a maze:** This problem is a classic application of **Reinforcement Learning**. An agent learns to make a sequence of decisions in an environment to maximize a cumulative reward.

*   **Input Data:** The input data for the robot (agent) would be its current state, which includes its position within the maze and potentially sensor readings (e.g., proximity to walls).
*   **Output:** The output is a policy that dictates the optimal action for the robot to take from any given state in the maze (e.g., move forward, turn left, turn right).
*   **Learning Signal/Objective:** The learning signal is a reward signal from the environment; positive rewards for reaching the goal, negative rewards for collisions or taking longer paths, and a cumulative reward to maximize over time to find the shortest path.
