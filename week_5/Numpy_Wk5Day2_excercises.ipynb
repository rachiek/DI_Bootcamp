# Exercise 2 : Type Conversion and Array Operations
Instructions
Convert a list [3.14, 2.17, 0, 1, 2] into a NumPy array and convert its data type to integer.

```python
a_list= [3.14, 2.17, 0, 1, 2]
converted_list = np.array(a_list, dtype=int)
print(converted_list)
print(f"array type: {converted_list.dtype}")
```

# Excercise 8: Simple Aggregate Funcs
Instructions
Find the sum and average of a given 1D array.

```python
arr = np.arange(10)

array_sum = np.sum(arr)
array_average = np.mean(arr)

print(f"Sum: {array_sum}")
print(f"Average: {array_average}")
```
