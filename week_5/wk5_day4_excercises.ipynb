# Excercise 1

Task: Explain why data visualization is important in data analysis. Task: Describe the purpose of a line graph in data visualization

 Raw data, especially large datasets, can be overwhelming and difficult to interpret. Visualizations transform this data into an easily digestible format, allowing for quick comprehension of trends, patterns, and outliers that might otherwise go unnoticed.
 Visual displays like charts and graphs make it simple to spot correlations, distributions, and shifts in data over time or across categories. This helps analysts draw meaningful conclusions.
 By presenting insights clearly and concisely, data visualizations empower stakeholders to make informed decisions faster. They provide a common ground for discussion and understanding among diverse audiences.
 Visuals are often more persuasive and memorable than tables of numbers. They enable analysts to tell a compelling story with their data, making complex findings accessible to both technical and non-technical audiences.
 Outliers or unusual data points that could indicate errors or significant events are much easier to identify visually than by scanning spreadsheets.

A line graph (or line chart) is primarily used to display quantitative data over a continuous interval or time period. Its main purposes include:

 This is its most common use. Line graphs are excellent for visualizing how a particular data point or variable changes over a series of time units (e.g., daily stock prices, monthly sales, yearly population growth).

You can plot multiple lines on the same graph to compare trends of different categories or groups over the same period, allowing for easy identification of similarities, differences, and crossovers.
Illustrating Relationships: When two variables are plotted, a line graph can reveal a relationship or correlation between them as they evolve over a continuous dimension.
Forecasting and Prediction: By observing past trends, line graphs can be used to infer future behavior, although this requires more sophisticated analytical methods.
In essence, a line graph is best suited when you want to emphasize changes and continuity in data points, especially when the x-axis represents time.



# Excercise 2

Objective: Create a simple line plot using Matplotlib that represents temperature variations over a week.
Tasks:
Use a list of temperature values for each day of the week (e.g., [72, 74, 76, 80, 82, 78, 75]).
Label the x-axis as “Day” and the y-axis as “Temperature (°F)”.
Add a title to the plot.
Display the plot.

```python
import matplotlib.pyplot as plt

temperatures = [72, 74, 76, 80, 82, 78, 75]
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

plt.plot(days, temperatures)
plt.xlabel('Day')
plt.ylabel('Temperature (°F)')
plt.title('Temperature Variation Over a Week')

plt.show()
```

# Excercise 3

Objective: Generate a bar chart using Matplotlib to visualize monthly sales data for a retail store.

# Excercise 5
Comparing Anxiety Levels Across Different Genders
Objective: Use a bar plot to compare the proportion of students experiencing anxiety across different genders.
Dataset Overview: The dataset includes columns: ‘Do you have Anxiety?’ and ‘Choose your gender’.
Tasks:
Import necessary libraries.
Use Seaborn to create a bar plot comparing anxiety levels across genders from the dataset df.
Customize the plot with an appropriate color palette and add a title.
Display the plot.

```python
# Convert Anxiety to a binary numeric column (1 for Yes, 0 for No) to compute proportions
mental_raw['Anxiety_Binary'] = mental_raw['Do you have Anxiety?'].map({'Yes': 1, 'No': 0})

# Create the bar plot
plt.figure(figsize=(6, 5))
sns.barplot(
    data=mental_raw,
    x='Choose your gender',
    y='Anxiety_Binary',
    palette='pastel',
    errorbar=None  # Set to None if you don't want error bars shown
)

# Customize titles and labels
plt.title('Proportion of Students with Anxiety Across Genders')
plt.xlabel('Gender')
plt.ylabel('Proportion with Anxiety')
plt.ylim(0, 1)

# Display the plot
plt.show()
```

# Exercise 6: Exploring the Relationship Between Age and Panic Attacks
Objective: Create a scatter plot to explore the relationship between students’ age and the occurrence of panic attacks.
Dataset Overview: The dataset includes columns: ‘Age’ and ‘Do you have Panic Attacks?’.
Tasks:
Import necessary libraries.
Convert panic attack responses to numeric values (e.g., Yes=1, No=0).
Use Seaborn’s scatterplot to create a scatter plot with ‘Age’ on the x-axis and numeric panic attack responses on the y-axis.
Customize the plot to improve readability by adding labels, a title, and adjusting point styles.
Display the plot.

```python
# Convert panic attack responses to numeric binary values (1 for Yes, 0 for No)
mental_raw['Panic_Attack_Binary'] = mental_raw['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

# Set up the figure
plt.figure(figsize=(8, 5))

# Create scatter plot
sns.scatterplot(
    data=mental_raw,
    x='Age',
    y='Panic_Attack_Binary',
    alpha=0.6,          # Adds transparency to see overlapping points
    s=70,               # Marker size
    color='darkred'
)

# Customize title and labels
plt.title('Relationship Between Student Age and Occurrence of Panic Attacks')
plt.xlabel('Age')
plt.ylabel('Panic Attack (1 = Yes, 0 = No)')
plt.yticks([0, 1], ['No (0)', 'Yes (1)'])
plt.grid(True, linestyle='--', alpha=0.5)

# Display the plot
plt.show()
```

```python
import numpy as np

# Ensure binary column exists
mental_raw['Panic_Attack_Binary'] = mental_raw['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

# Add slight random jitter to y so points don't stack directly over each other
y_jittered = mental_raw['Panic_Attack_Binary'] + np.random.uniform(-0.06, 0.06, size=len(mental_raw))

plt.figure(figsize=(9, 5))

# Strictly uses sns.scatterplot as required by the prompt
sns.scatterplot(
    x=mental_raw['Age'],
    y=y_jittered,
    alpha=0.5,
    s=80,
    color='crimson',
    edgecolor='black',
    linewidth=0.5
)

plt.title('Relationship Between Student Age and Panic Attacks', fontsize=13)
plt.xlabel('Age', fontsize=11)
plt.ylabel('Panic Attack', fontsize=11)
plt.yticks([0, 1], ['No (0)', 'Yes (1)'])
plt.ylim(-0.2, 1.2)
plt.xticks(sorted(mental_raw['Age'].dropna().unique()))
plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.show()
```
