```python
#Excercise 2
import numpy as np
from scipy import stats

data = [12, 15, 13, 12, 18, 20, 22, 21]

# Convert to numpy array for easier calculations
data_np = np.array(data)

# Calculate statistics using NumPy
mean_val = stats.tmean(data_np)
median_val = np.median(data_np)
variance_val = stats.tvar(data_np)
std_dev_val = stats.tstd(data_np)

print(f"Data: {data}")
print(f"Mean: {mean_val}")
print(f"Median: {median_val}")
print(f"Variance: {variance_val}")
print(f"Standard Deviation: {std_dev_val}")
```

```python
#Excercise 3
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Define mean and standard deviation
mean = 50
std_dev = 10

# Generate a range of values for the x-axis
x = np.linspace(mean - 4 * std_dev, mean + 4 * std_dev, 500)

# Calculate the probability density function (PDF) for the normal distribution
pdf = norm.pdf(x, loc=mean, scale=std_dev)

# Plot the distribution
plt.figure(figsize=(10, 6))
plt.plot(x, pdf, label=f'Mean = {mean}, Std Dev = {std_dev}')
plt.title('Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Excercise 4: T-Test Application
import numpy as np
from scipy import stats

# Generate two sets of randomly generated data
data1 = np.random.normal(50, 10, 100) # Mean 50, Std Dev 10, 100 samples
data2 = np.random.normal(60, 10, 100) # Mean 60, Std Dev 10, 100 samples

# Perform an independent samples T-test
t_statistic, p_value = stats.ttest_ind(data1, data2)

print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpret the results
alpha = 0.05 # Significance level
if p_value < alpha:
    print("\nInterpretation: Since the P-value is less than 0.05, we reject the null hypothesis.")
    print("This suggests there is a statistically significant difference between the means of the two datasets.")
else:
    print("\nInterpretation: Since the P-value is greater than 0.05, we fail to reject the null hypothesis.")
    print("This suggests there is no statistically significant difference between the means of the two datasets.")
```

```python
#Excercise 5
#Define the datasets
house_sizes = np.array([50, 70, 80, 100, 120]) # in square meters
house_prices = np.array([150000, 200000, 210000, 250000, 280000]) # in currency units

# Perform linear regression
slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)

print(f"Slope of the regression line: {slope:.2f}")
print(f"Intercept of the regression line: {intercept:.2f}")

# Predict the price of a house that is 90 square meters
predicted_price = slope * 90 + intercept
print(f"Predicted price for a 90 square meter house: {predicted_price:.2f}")

# Interpret the meaning of the slope
print(f"\nInterpretation of the slope: For every one square meter increase in house size, the house price is predicted to increase by {slope:.2f} currency units.")
```

```python
#Excercise 6
from scipy import stats

# Define the datasets
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

# Perform ANOVA test
f_statistic, p_value_anova = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print(f"F-value: {f_statistic:.4f}")
print(f"P-value: {p_value_anova:.4f}")

# Interpret the results based on P-value
alpha = 0.05 # Significance level

if p_value_anova < alpha:
    print("\nInterpretation: Since the P-value is less than 0.05, we reject the null hypothesis.")
    print("This suggests that there is a statistically significant difference in plant growth due to the different fertilizers.")
else:
    print("\nInterpretation: Since the P-value is greater than 0.05, we fail to reject the null hypothesis.")
    print("This suggests that there is no statistically significant difference in plant growth due to the different fertilizers.")
    print("If the P-value were greater than 0.05, it would mean that any observed differences in plant growth among the fertilizer groups are likely due to random chance, and we cannot conclude that the fertilizers have a significantly different effect.")
```

```python
#Excercise 7
from scipy.stats import binom

# Parameters for the binomial distribution
n = 10  # Number of coin flips (trials)
p = 0.5 # Probability of getting a head (success)
k = 5   # Number of desired heads (successes)

# Calculate the probability of getting exactly k successes
probability = binom.pmf(k, n, p)

print(f"Number of trials (n): {n}")
print(f"Probability of success (p): {p}")
print(f"Number of desired successes (k): {k}")
print(f"Probability of getting exactly {k} heads in {n} flips: {probability:.4f}")
```

```python
#Excercise 8
import pandas as pd
from scipy.stats import pearsonr, spearmanr

# Define the dataset
data = pd.DataFrame({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})

# Extract the variables
age = data['age']
income = data['income']

# Calculate Pearson correlation coefficient
pearson_corr, pearson_p_value = pearsonr(age, income)
print(f"Pearson Correlation Coefficient: {pearson_corr:.4f} (P-value: {pearson_p_value:.4f})")

# Calculate Spearman correlation coefficient
spearman_corr, spearman_p_value = spearmanr(age, income)
print(f"Spearman Correlation Coefficient: {spearman_corr:.4f} (P-value: {spearman_p_value:.4f})")

print("\nInterpretation: Both Pearson and Spearman correlation coefficients measure the strength and direction of a monotonic relationship between two variables. Pearson measures linear relationships, while Spearman measures monotonic (not necessarily linear) relationships. A value of 1 indicates a perfect positive correlation.")
```
